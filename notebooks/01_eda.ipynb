# 01 - Exploratory Data Analysis (EDA)
Smart Waste Classification project.
This notebook discovers the local dataset, runs validation, inspects image dimensions and class distributions, and outputs balanced loss weights.

```python
from waste_classifier.data import discover_classes, run_eda
```

```python
# 1. Dynamic class discovery
classes = discover_classes("data/raw")
print(f"Discovered {len(classes)} classes: {classes}")
```

```python
# 2. Run comprehensive inspection and EDA reporting
stats = run_eda(data_dir="data/raw", output_dir="reports")
print(f"Total valid images: {stats.valid_images_count}")
print(f"Imbalance ratio: {stats.imbalance_ratio}")
for cls, cs in stats.class_summaries.items():
    print(f"{cls:<12}: {cs.count} images ({cs.percentage}%) | Weight: {cs.class_weight}")
```
